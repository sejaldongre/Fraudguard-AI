# FraudGuard AI — Data Preprocessing

## Objective

Prepare the credit-card transaction data for machine-learning model training.

This notebook will:

- Handle duplicate records
- Separate features and target
- Split data into training and testing sets
- Scale appropriate numerical features
- Prepare a leakage-safe preprocessing pipeline
- Preserve the original class distribution for baseline modeling

# Preprocessing Summary

The following preprocessing steps were performed:

1. Exact duplicate rows were removed.
2. The target variable (`Class`) was separated from the input features.
3. The dataset was divided into training and testing sets using an 80/20 split.
4. Stratified splitting was used to preserve the fraud/legitimate class distribution.
5. `Time` and `Amount` were standardized using statistics learned only from the training data.
6. The test set was not used when fitting preprocessing transformations.
7. No class-balancing technique was applied yet so that a baseline model can first be established.

These decisions help reduce data leakage and provide a reliable foundation for model evaluation.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("../data/raw/creditcard.csv")
df.head()
```

```python
df.shape
```

```python
duplicate_count=df.duplicated().sum()
print("Duplicate_counts:",duplicate_count)
```

```python
df=df.drop_duplicates().copy()
print(df)
```

```python
print("shape after removing duplicates:",df.shape)
print("number of duplicates after removing duplicates:",df.duplicated().sum())
```

```python
df["Class"].value_counts()
```

```python
df["Class"].value_counts(normalize=True)*100
```

```python
X=df.drop(columns=["Class"])
y=df["Class"]

print(X.head())
print(y.head())
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
    
)
```

```python
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
print("Training class distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
scaler = StandardScaler()
```

```python
X_train = X_train.copy()
X_test = X_test.copy()

X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)
```

```python
X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]])
```
